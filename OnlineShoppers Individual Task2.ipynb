```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC

from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

# Cross-validation and reliability of model evaluation


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import (
    StratifiedKFold,
    cross_validate
)

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

from sklearn.model_selection import learning_curve
import numpy as np
import matplotlib.pyplot as plt
```

```python
online_data = pd.read_csv("online_shoppers_intention.csv")

online_data.head()
```

```python
print("Number of rows and columns:", online_data.shape)

print("\nColumn names:")
print(online_data.columns)

print("\nDataset information:")
online_data.info()
```

```python
print("Missing values:")
print(online_data.isnull().sum())
```

```python
print("Number of duplicate rows:", online_data.duplicated().sum())
```

```python
online_data = online_data.drop_duplicates()

print("Shape after removing duplicates:", online_data.shape)
```

```python
print(online_data['Revenue'].value_counts())
print(online_data['Revenue'].value_counts(normalize=True) * 100)
```

```python
online_data['Revenue'].value_counts().plot(kind='bar')

plt.title("Online Shopping Revenue Distribution")
plt.xlabel("Revenue")
plt.ylabel("Number of Sessions")

plt.show()
```

```python
online_data['Revenue'] = online_data['Revenue'].map({
    False: 0,
    True: 1
})
```

```python
X_online = online_data.drop('Revenue', axis=1)

y_online = online_data['Revenue']
```

```python
print("Features:")
print(X_online.columns)

print("\nTarget:")
print(y_online.name)
```

```python
X_online = pd.get_dummies(
    X_online,
    drop_first=True
)
```

```python
print("Number of features after encoding:", X_online.shape[1])

X_online.head()
```

```python
X_train_online, X_test_online, y_train_online, y_test_online = train_test_split(
    X_online,
    y_online,
    test_size=0.25,
    random_state=0,
    stratify=y_online
)

print("Training data:", X_train_online.shape)
print("Testing data:", X_test_online.shape)
```

```python
random_forest_online = RandomForestClassifier(
    n_estimators=100,
    random_state=0
)

random_forest_online.fit(
    X_train_online,
    y_train_online
)
```

```python
y_pred_rf_online = random_forest_online.predict(
    X_test_online
)
```

```python
rf_online_accuracy = accuracy_score(
    y_test_online,
    y_pred_rf_online
)

rf_online_precision = precision_score(
    y_test_online,
    y_pred_rf_online
)

rf_online_recall = recall_score(
    y_test_online,
    y_pred_rf_online
)

rf_online_f1 = f1_score(
    y_test_online,
    y_pred_rf_online
)

print("Random Forest Accuracy:", rf_online_accuracy)
print("Random Forest Precision:", rf_online_precision)
print("Random Forest Recall:", rf_online_recall)
print("Random Forest F1-score:", rf_online_f1)
```

```python
print("Random Forest Classification Report:")
print(
    classification_report(
        y_test_online,
        y_pred_rf_online
    )
)
```

```python
rf_online_cm = confusion_matrix(
    y_test_online,
    y_pred_rf_online
)

print("Random Forest Confusion Matrix:")
print(rf_online_cm)
```

```python
plt.imshow(rf_online_cm)

plt.title("Online Shoppers - Random Forest Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.colorbar()

plt.show()
```

```python
online_feature_importance = pd.Series(
    random_forest_online.feature_importances_,
    index=X_online.columns
)

online_feature_importance = online_feature_importance.sort_values(
    ascending=False
)

print("Top 10 Important Features:")
print(
    online_feature_importance.head(10)
)
```

```python
# SVM
scaler_online = StandardScaler()

X_train_online_scaled = scaler_online.fit_transform(
    X_train_online
)

X_test_online_scaled = scaler_online.transform(
    X_test_online
)
```

```python
svm_online = SVC(
    kernel='rbf',
    random_state=0
)

svm_online.fit(
    X_train_online_scaled,
    y_train_online
)
```

```python
y_pred_svm_online = svm_online.predict(
    X_test_online_scaled
)
```

```python
svm_online_accuracy = accuracy_score(
    y_test_online,
    y_pred_svm_online
)

svm_online_precision = precision_score(
    y_test_online,
    y_pred_svm_online
)

svm_online_recall = recall_score(
    y_test_online,
    y_pred_svm_online
)

svm_online_f1 = f1_score(
    y_test_online,
    y_pred_svm_online
)

print("SVM Accuracy:", svm_online_accuracy)
print("SVM Precision:", svm_online_precision)
print("SVM Recall:", svm_online_recall)
print("SVM F1-score:", svm_online_f1)
```

```python
print("SVM Classification Report:")
print(
    classification_report(
        y_test_online,
        y_pred_svm_online
    )
)
```

```python
svm_online_cm = confusion_matrix(
    y_test_online,
    y_pred_svm_online
)

print("SVM Confusion Matrix:")
print(svm_online_cm)
```

```python
plt.imshow(svm_online_cm)

plt.title("Online Shoppers - SVM Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.colorbar()

plt.show()
```

```python
online_results = pd.DataFrame({
    'Model': ['Random Forest', 'SVM'],
    'Accuracy': [
        rf_online_accuracy,
        svm_online_accuracy
    ],
    'Precision': [
        rf_online_precision,
        svm_online_precision
    ],
    'Recall': [
        rf_online_recall,
        svm_online_recall
    ],
    'F1-score': [
        rf_online_f1,
        svm_online_f1
    ]
})

print(online_results)
```

```python
online_results.set_index('Model')[[
    'Accuracy',
    'Precision',
    'Recall',
    'F1-score'
]].plot(kind='bar')

plt.title("Online Shoppers Model Performance")
plt.xlabel("Model")
plt.ylabel("Score")
plt.xticks(rotation=0)

plt.show()
```

```python
# Evaluation metrics for cross-validation

scoring = {
    'accuracy': 'accuracy',
    'precision': 'precision',
    'recall': 'recall',
    'f1': 'f1'
}

# 5-fold cross-validation - Online Shoppers


cv_online = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=0
)

# Random Forest

cv_results_rf_online = cross_validate(
    random_forest_online,
    X_train_online,
    y_train_online,
    cv=cv_online,
    scoring=scoring,
    n_jobs=-1
)

print("ONLINE SHOPPERS - RANDOM FOREST")
print("--------------------------------")

for metric in scoring.keys():
    scores = cv_results_rf_online[f'test_{metric}']
    
    print(
        f"{metric.capitalize()}: "
        f"{scores.mean():.4f} ± {scores.std():.4f}"
    )
```

```python

# SVM with scaling inside the pipeline


svm_pipeline_online = Pipeline([
    ('scaler', StandardScaler()),
    ('svm', SVC(
        kernel='rbf',
        random_state=0
    ))
])

cv_results_svm_online = cross_validate(
    svm_pipeline_online,
    X_train_online,
    y_train_online,
    cv=cv_online,
    scoring=scoring,
    n_jobs=-1
)

print("ONLINE SHOPPERS - SVM")
print("---------------------")

for metric in scoring.keys():
    scores = cv_results_svm_online[f'test_{metric}']
    
    print(
        f"{metric.capitalize()}: "
        f"{scores.mean():.4f} ± {scores.std():.4f}"
    )
```

```python
# Cross-validation summary table - Online Shoppers


online_cv_summary = pd.DataFrame({
    'Model': ['Random Forest', 'SVM'],
    
    'Accuracy Mean': [
        cv_results_rf_online['test_accuracy'].mean(),
        cv_results_svm_online['test_accuracy'].mean()
    ],
    
    'Accuracy SD': [
        cv_results_rf_online['test_accuracy'].std(),
        cv_results_svm_online['test_accuracy'].std()
    ],
    
    'Precision Mean': [
        cv_results_rf_online['test_precision'].mean(),
        cv_results_svm_online['test_precision'].mean()
    ],
    
    'Recall Mean': [
        cv_results_rf_online['test_recall'].mean(),
        cv_results_svm_online['test_recall'].mean()
    ],
    
    'F1 Mean': [
        cv_results_rf_online['test_f1'].mean(),
        cv_results_svm_online['test_f1'].mean()
    ],
    
    'F1 SD': [
        cv_results_rf_online['test_f1'].std(),
        cv_results_svm_online['test_f1'].std()
    ]
})

print(online_cv_summary.round(4))
```

```python

# Learning Curve - Online Shoppers Random Forest

train_sizes_online, train_scores_online, validation_scores_online = learning_curve(
    estimator=random_forest_online,
    X=X_train_online,
    y=y_train_online,
    cv=cv_online,
    scoring='f1',
    train_sizes=np.linspace(0.1, 1.0, 6),
    n_jobs=-1
)

# Calculate mean and standard deviation
train_mean_online = train_scores_online.mean(axis=1)
train_std_online = train_scores_online.std(axis=1)

validation_mean_online = validation_scores_online.mean(axis=1)
validation_std_online = validation_scores_online.std(axis=1)

# Print values
learning_curve_online = pd.DataFrame({
    'Training Samples': train_sizes_online,
    'Training F1 Mean': train_mean_online,
    'Validation F1 Mean': validation_mean_online
})

print(learning_curve_online.round(4))
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    train_sizes_online,
    train_mean_online,
    marker='o',
    label='Training F1'
)

plt.plot(
    train_sizes_online,
    validation_mean_online,
    marker='o',
    label='Validation F1'
)

plt.fill_between(
    train_sizes_online,
    train_mean_online - train_std_online,
    train_mean_online + train_std_online,
    alpha=0.15
)

plt.fill_between(
    train_sizes_online,
    validation_mean_online - validation_std_online,
    validation_mean_online + validation_std_online,
    alpha=0.15
)

plt.xlabel('Number of Training Samples')
plt.ylabel('F1-score')
plt.title('Learning Curve - Online Shoppers Random Forest')
plt.legend()
plt.grid(True)

plt.show()
```

```python

# Fairness analysis - Online Shoppers - by Visitor Type

from fairlearn.metrics import (
    MetricFrame,
    selection_rate,
    true_positive_rate,
    false_positive_rate,
    false_negative_rate,
    equalized_odds_difference
)

# Reload original Online Shoppers data 
online_fairness_data = pd.read_csv("online_shoppers_intention.csv")
online_fairness_data = online_fairness_data.drop_duplicates()

online_fairness_data['Revenue'] = online_fairness_data['Revenue'].map({
    False: 0,
    True: 1
})

X_online_raw = online_fairness_data.drop('Revenue', axis=1)
y_online_raw = online_fairness_data['Revenue']

# Recreate exactly the same train/test split as the main analysis
X_online_raw_train, X_online_raw_test, y_online_raw_train, y_online_raw_test = train_test_split(
    X_online_raw,
    y_online_raw,
    test_size=0.25,
    random_state=0,
    stratify=y_online_raw
)

print("Raw test data shape:", X_online_raw_test.shape)
print("Encoded test data shape:", X_test_online.shape)

# Check indices match before using VisitorType as sensitive feature

print("Indices match:", X_online_raw_test.index.equals(X_test_online.index))

# Visitor type groups (used as-is; already categorical)

visitor_type_test = X_online_raw_test['VisitorType']

print(visitor_type_test.value_counts())



# Random Forest predictions (already computed model, encoded test set)
y_pred_rf_online_fair = random_forest_online.predict(X_test_online)

print("Number of predictions:", len(y_pred_rf_online_fair))
print("Number of actual values:", len(y_test_online))


# Fairlearn MetricFrame - Online Shoppers by Visitor Type

fairness_metrics_online = {
    'Selection Rate': selection_rate,
    'True Positive Rate': true_positive_rate,
    'False Positive Rate': false_positive_rate,
    'False Negative Rate': false_negative_rate
}

metric_frame_online = MetricFrame(
    metrics=fairness_metrics_online,
    y_true=y_test_online,
    y_pred=y_pred_rf_online_fair,
    sensitive_features=visitor_type_test
)

print("FAIRNESS RESULTS BY VISITOR TYPE")
print("==================================")

print(
    metric_frame_online.by_group.round(4)
)


group_counts_online = visitor_type_test.value_counts()

fairness_table_online = metric_frame_online.by_group.copy()
fairness_table_online['Count'] = group_counts_online

print(fairness_table_online.round(4))

# Differences between visitor type groups

fairness_difference_online = metric_frame_online.difference()

print("Maximum difference between visitor type groups:")
print(fairness_difference_online.round(4))

# Equalized Odds Difference - Online Shoppers

eo_difference_online = equalized_odds_difference(
    y_true=y_test_online,
    y_pred=y_pred_rf_online_fair,
    sensitive_features=visitor_type_test
)

print(
    "Equalized Odds Difference:",
    round(eo_difference_online, 4)
)


# Fairness plot - Online Shoppers

metric_frame_online.by_group[
    [
        'Selection Rate',
        'True Positive Rate',
        'False Positive Rate',
        'False Negative Rate'
    ]
].plot(kind='bar', figsize=(10, 6))

plt.title('Online Shoppers Random Forest Performance by Visitor Type')

plt.xlabel('Visitor Type')
plt.ylabel('Metric Value')

plt.xticks(rotation=0)

plt.legend(
    bbox_to_anchor=(1.05, 1),
    loc='upper left'
)

plt.tight_layout()
plt.show()
```
